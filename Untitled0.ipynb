```python
def fibonacci(n):
    deret = []
    a, b = 1, 1
    for i in range(n):
        deret.append(str(a))
        a, b = b, a + b
    return ", ".join(deret)

def main():
    while True:
        print("\nNim Genap")
        print("menu pilihan")
        print("1. Barisan Fibonacci")
        print("2. M Kali N")
        print("0. Keluar")

        pilihan = input("Pilih Menu : ")

        if pilihan == '1':
            n_suku = int(input("Masukkan Jumlah Suku : "))
            hasil = fibonacci(n_suku)
            print(f"\nbarisan fibonacci sebanyak {n_suku} suku :")
            print(hasil)

        elif pilihan == '2':
            m = int(input("Masukkan Suatu Bilangan Bulat : "))
            n = int(input("Masukkan Suatu Bilangan Pengali : "))
            hasil_kali = m * n
            print(f"\n{m} x {n} = {hasil_kali}")

        elif pilihan == '0':
            print("Keluar dari program...")
            break
        else:
            print("Pilihan tidak valid, silakan coba lagi.")

if __name__ == "__main__":
    main()
```
